```python
import pandas as pd
from pathlib import Path

# 🔧 Set your directory path here
data_dir = Path(r"D:\FYP_DEV\app\data\extracted\final")  # change this to your folder path

# 🧾 Find all CSV files in the folder (non-recursive)
csv_files = list(data_dir.glob("*.csv"))

# 🧠 Load and concatenate all CSVs
df_list = [pd.read_csv(f) for f in csv_files]
combined_df = pd.concat(df_list, ignore_index=True)

# 📊 Display total number of rows and CSV count
print(f"Loaded {len(csv_files)} CSV files.")
print(f"Total rows across all files: {len(combined_df)}")

# (Optional) Show first few rows to verify
print(combined_df.head())
```

```python
import os
import pandas as pd

# --- User inputs ---
input_csv  = r"D:\FYP_DEV\AIMchat1.csv"     # Path to input CSV
output_csv = r"D:\FYP_DEV\AIMchat2.csv"  # Path to output CSV
num_lines  = 20                             # Number of lines (from bottom) to copy

# --- Read bottom N lines from input file ---
df_input = pd.read_csv(input_csv)
bottom_rows = df_input.tail(num_lines)

# --- Append to destination CSV ---
# If file exists, open in append mode (no header)
# If not, write header
file_exists = os.path.exists(output_csv)

bottom_rows.to_csv(
    output_csv,
    mode="a",           # append mode
    header=not file_exists,  # write header only if new file
    index=False
)

print(f"✅ Appended last {num_lines} rows from:\n{input_csv}\n→ {output_csv}")
```
